# Introduction

The following notebook calculates some urban indicators for Tampere and Helsinki. In Section [2](#Methods-and-data), data and methods used in the study will be introduced in more detail. In Section [3](#Bicycle-network-analysis), the bicycle networks of the cities will be analyzed both in terms of traditional statistics and in terms of spatial properties of the network using data from OpenStreetMap.org (later **OSM**). The spatial properties of the network can be divided into geometric and topological properties, where geometric properties only consider the geometry of the objects and topological properties tell something about the relationships between the nodes and edges in the network. Geometric properties include attributes such as length of the edges and the area of their total hull whereas topological properties include attributes such as centrality and node degree. Section [4](#Urban-growth-in-Tampere-and-Helsinki-between-2016-and-2020) will concentrate on analyzing urban growth in Tampere and Helsinki. Urban growth is an interesting indicator to compare between these cities, as both are rapidly developing cities, some of the fastest growing cities in Finland.

The aim of the study is to compare the bicycle network of the cities to see if there are some interesting differences between the two networks. Some interesting measures for the comparison, such as amount and geometries of network components, will be calculated for both cities. The purpose of examining the network components is to see if there are some discontinuities in the networks. Another aim of the study is to compare and visualize the urban growth of the cities from 2016 to 2020 using data from **Statistics Finland**. The growth will be visualized with maps and graphs to detect patterns in the data.

## Bicycle network analysis data and methods

Data from OpenStreetMap.org (OSM) will be used for analyzing the bicycle networks of the two cities. The data is fetched using `osmnx` which is a mature Python library for fetching data from OSM via Nominatim geocoder and Overpass API. `pandas` and `geopandas` Python libraries will be used for calculating traditional and spatial statistics for the data whereas `osmnx` and `networkx` libraries will be used for collecting topological information from the spatial networks.

The spatial network fetched using `osmnx` contains a column `highway`, which contains a single or multiple categories the edge of the network belongs to. The value counts for the categories will be calculated both in terms of mutually inexclusive categories and mutually exclusive sets of categories and joined together using the categories or the set of categories as keys. The counts are then printed as tables and shown as bar charts. 

The basic stats are calculated for both networks of the two cities and compared using tables and plots. The networks are then changed into undirected networks and connected components of the networks are calculated using `NetworkX` library. The labels of the connected components are used to label each edge by its subgraph component number and visualized on the map by classifying by the size of the connected component. Additionally, the ten most common roadlink types of both cities are visualized on a map. Finally, all the plots are analyzed using visual analysis methods.

The code for the analysis is organized as a separate Python module `street_network_metrics` which contains functions that are used to compute network statistics for the two cities. Additionally, some generally applicable functions can be found in `utils` module. Visualizing is performed mostly in the notebook itself but some have been organized as functions in either of the modules depending on whether it's a generally applicable function or just used for a single purpose of the respective analysis.   

## Urban growth analysis data and methods

Data from Statistics Finland will be used for studying the urban growth of the cities. The data sets used for the comparison are the *Population grid data 1 km x 1 km* and *Municipalities, key figures* data sets from 2016 to 2020. The data is downloaded from *Paituli.csc.fi* portal using their WFS service, analyzed with `pandas` and `geopandas` and plotted using `matplotlib`.

The municipal level data, *Municipalities, key figures*, is joined as one DataFrame where each row index marks the year of the statistics, the difference between consecutive years is calculated both in absolute and relative terms and then plotted as a line chart for each demographics of the population: Males, 15--64-year-olds, 0--14-year-olds, over 65-year-olds, all population and females. 

After plotting the line chart, the GeoDataFrames for the grid dataset from 2016 to 2020 is combined into one GeoDataFrame, the absolute and relative change of population is calculated for each demographic group from 2016 to 2020 by using values from first and the last year (2016 and 2020) of the timespan. After calculating the change from 2016 to 2020, the change values are then classified using `mapclassify` into uniform classes by using the data from both cities as the input for the classifier. Change in the population of each demographic group is then plotted as a series of chropleth maps, one figure for each demographic in the data. Plots are then analyzed with visual geospatial data analysis methods.

# Reflection

The goals of the bicycle networks analysis was to examine if there are some interesting differences between the bicycle networks of the two cities or some discontinuities in the network. The main difference found was the slightly different spatial distribution of 10 most common roadlink types for the network. I.e. Helsinki has very clear main routes as  explained in section [5.1.2.](#Spatial-distribution-of-roadlink-types) whereas Tampere has more uniform distribution of roadlink types in the urban centre. Additionally, network component analysis discovered (see section [5.1.4.](#Network-component-analysis)) that Helsinki has far more disconnected bicycle network components. 

It can not be said for certain if the discontinuities are actual gaps in the bicycle network of the city or just irregularities in the classification of OpenStreetMap network edges (`way`). Likewise, it can not be said for certain whether the clear main routes of Helsinki seen in section [5.1.2.](#Spatial-distribution-of-roadlink-types) is real phenomenon or just an artefact of `way` classification in the underlying OSM data. When examining the data the first time, I saw some gaps in the data in the bike lanes of Hämeentie in Helsinki, that should not be there . There could easily be similar gaps in other parts of the city as well, which would, in part, explain the significant difference between the number of subgraphs (components) between the two cities. Although it also raises a question why the data for Tampere does not contain as many gaps like that.  

In section [5.2.2.](#Distribution-of-population-change) I attempted to interpret the spatial distribution of population change for different demographic groups. It was too easy to make too broad conclusions from the maps using the absolute change maps. However, nearly always the relative change map was not nearly as conclusive. Likewise, by comparing the results in that section to the municipal level results in section [5.2.1.](#Population-change-on-municipal-level) it can be said that some of the interpretations about change in population are perhaps too broad. 

Although it first seemed like a good idea to classify data for both cities with the same classifier to get uniform class boundaries, it was not quite as good an idea as I initially thought. Even though the data was easier to compare between the cities, it made reading the details of a single city from a map a bit difficult as the difference in scale of the data for both cities was too large. For example, the maximum value of some column was several times larger for Helsinki than for Tampere which made the colors too similar for some classes. Part of the idea was to make negative change distinguishable from positive change but that turned out to be more tricky than thought. 

Other alternatives for uniform classification across cities would have been to use classification based on one city and then reclassify using that classification or use the same classifier for both cities but arbitrary colors for the visualization. The disadvantage of those approaches would have been that the data would not be easy to compare across the cities or the colors would be completely meaningless without looking at the legend. In the end, the classification and visualization method chosen for the task was a decent compromise between data comparability and interpretability.
